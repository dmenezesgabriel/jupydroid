```python
# Plants three kernelspecs the way `ipykernel install` lays one out, reduced to its kernel.json,
# and prints the server's own search list: the kernel inherits the server's environment, so
# jupyter_path("kernels") here is the list jupyter_server walks.
import json, sys, pathlib
from jupyter_core.paths import jupyter_path
print(*jupyter_path("kernels"), sep="\n")
home_dir = pathlib.Path(jupyter_path("kernels")[0])
local_dir = pathlib.Path("/usr/local/share/jupyter/kernels")
def plant(root, name, display):
    spec_dir = root / name
    spec_dir.mkdir(parents=True, exist_ok=True)
    (spec_dir / "kernel.json").write_text(json.dumps({
        "argv": [sys.executable, "-m", "jupydroid_kernel", "-f", "{connection_file}"],
        "display_name": display,
        "language": "python",
    }))
plant(home_dir, "probe-home", "Probe home")
plant(local_dir, "probe-local", "Probe local")
plant(home_dir, "probe-clash", "Probe clash from home")
plant(local_dir, "probe-clash", "Probe clash from local")
print("specs_" + "planted")
```
