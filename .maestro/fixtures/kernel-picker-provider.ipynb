```
# Plants a second kernelspec where the running server's KernelSpecManager will
# find it - the user-level act `python -m ipykernel install --user` performs,
# reduced to the kernel.json it would write. jupyter_path() reports the server's
# own search list (the kernel inherits its environment), and the dir holding the
# shipped `python3` spec is the one the next GET /api/kernelspecs re-scans.
import json, sys, pathlib
from jupyter_core.paths import jupyter_path
specs_root = next(
    pathlib.Path(d) for d in jupyter_path("kernels")
    if (pathlib.Path(d) / "python3" / "kernel.json").exists()
)
spec_dir = specs_root / "echo-kernel"
spec_dir.mkdir(parents=True, exist_ok=True)
(spec_dir / "kernel.json").write_text(json.dumps({
    "argv": [sys.executable, "-m", "jupydroid_kernel",
             "-f", "{connection_file}", "--log-level=DEBUG"],
    "display_name": "Echo Python",
    "language": "python",
    "env": {"PYTHONFAULTHANDLER": "1", "PICKER_KERNEL": "echo"},
    "metadata": {"debugger": True},
}))
print("spec_" + "planted")
```
