```
import struct, zlib
from IPython.display import Image, display

def striped_png(width, height):
    """A 1-bit greyscale PNG of black and white bars, each 800 px wide."""
    def chunk(kind, body):
        return struct.pack('>I', len(body)) + kind + body + struct.pack('>I', zlib.crc32(kind + body))
    bars = (b'\x00' * 100 + b'\xff' * 100) * (width // 1600)
    header = struct.pack('>IIBBBBB', width, height, 1, 0, 0, 0, 0)
    return (b'\x89PNG\r\n\x1a\n' + chunk(b'IHDR', header)
            + chunk(b'IDAT', zlib.compress((b'\x00' + bars) * height, 9)) + chunk(b'IEND', b''))

# 12800 x 3200 px decode to 163 840 000 bytes at four bytes a pixel, past the 150 MiB a canvas draws.
display(Image(striped_png(12800, 3200)))
print(6 * 7)
```
