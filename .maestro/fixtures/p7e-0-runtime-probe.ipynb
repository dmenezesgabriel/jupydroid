```
P = "PROBE" + "-1"

import sys, os
print(P, "executable:", sys.executable)
print(P, "prefix:", sys.prefix)
print(P, "base_prefix:", sys.base_prefix)
print(P, "version:", sys.version.replace("\n", " "))
print(P, "argv0:", sys.argv[0] if sys.argv else "")
print(P, "executable_exists:", os.path.exists(sys.executable))
print(P, "executable_x_ok:", os.access(sys.executable, os.X_OK))
print(P, "PATH:", os.environ.get("PATH", ""))
print(P, "HOME:", os.environ.get("HOME", ""))
```

```
P = "PROBE" + "-3"

import socket, os
print(P, "/etc/resolv.conf exists:", os.path.exists("/etc/resolv.conf"))
try:
    print(P, "/etc/resolv.conf:", open("/etc/resolv.conf").read().strip().replace("\n", " | "))
except Exception as e:
    print(P, "resolv read failed:", type(e).__name__, e)
try:
    print(P, "getaddrinfo:", socket.getaddrinfo("pypi.org", 443)[:2])
except Exception as e:
    print(P, "getaddrinfo failed:", type(e).__name__, e)
```

```
P = "PROBE" + "-4"

import ssl, os
print(P, "default_verify_paths:", ssl.get_default_verify_paths())
for p in ("/etc/ssl/certs/ca-certificates.crt", "/etc/ssl/certs"):
    print(P, "exists", p + ":", os.path.exists(p))
try:
    print(P, "cert_store_stats:", ssl.create_default_context().cert_store_stats())
except Exception as e:
    print(P, "context failed:", type(e).__name__, e)
```

```
P = "PROBE" + "-5"

import urllib.request
try:
    with urllib.request.urlopen("https://pypi.org/simple/", timeout=30) as r:
        body = r.read(200)
    print(P, "status:", r.status, "first bytes:", body[:80])
except Exception as e:
    print(P, "failed:", type(e).__name__, e)
```
