```
import numpy as np

samples = np.random.default_rng(7).normal(size=10_000)
print(f"mean {samples.mean():.3f}, std {samples.std():.3f}")
```

```
import matplotlib.pyplot as plt

counts, edges, _ = plt.hist(samples, bins=40, color="#4C8BF5", edgecolor="white")
plt.title("10,000 samples from a normal distribution")
plt.show()
```
