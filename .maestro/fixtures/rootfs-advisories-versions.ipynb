```python
import importlib.metadata
apk = {}
for record in open('/lib/apk/db/installed').read().split('\n\n'):
    fields = dict(line.split(':', 1) for line in record.splitlines() if line[1:2] == ':')
    if fields.get('P') in ('libcrypto3', 'libssl3', 'zlib'):
        apk[fields['P']] = fields['V']
pip = {name: importlib.metadata.version(name) for name in ('certifi', 'jupyter_server', 'tornado')}
print(' '.join(f'{k}={v}' for k, v in sorted(apk.items()) + list(pip.items())))
```
