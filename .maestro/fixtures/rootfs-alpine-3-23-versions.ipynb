```python
import importlib.metadata
apk = {}
for record in open('/lib/apk/db/installed').read().split('\n\n'):
    fields = dict(line.split(':', 1) for line in record.splitlines() if line[1:2] == ':')
    if fields.get('P') == 'zlib':
        apk['zlib'] = fields['V']
names = ('dnspython', 'opentelemetry-api', 'soupsieve', 'tornado')
pip = {name: importlib.metadata.version(name) for name in names}
print(' '.join(f'{k}={v}' for k, v in list(apk.items()) + list(pip.items())))
```
