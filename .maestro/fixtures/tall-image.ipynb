```
import struct, zlib
from IPython.display import Image, display

def chunk(kind, body):
    return struct.pack('>I', len(body)) + kind + body + struct.pack('>I', zlib.crc32(kind + body))

# 300 x 60000 px of 1-bit grey in bands 50 px tall: past the canvas floor, so decoded at half size.
band = lambda ink: (b'\x00' + ink * 38) * 50
rows = zlib.compress((band(b'\x00') + band(b'\xff')) * 600, 9)
header = struct.pack('>IIBBBBB', 300, 60000, 1, 0, 0, 0, 0)
print(6 * 7)
display(Image(b'\x89PNG\r\n\x1a\n' + chunk(b'IHDR', header) + chunk(b'IDAT', rows) + chunk(b'IEND', b'')))
```
