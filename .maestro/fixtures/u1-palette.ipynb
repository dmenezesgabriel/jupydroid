# Palette check

**Bold** and *italic* carry no colour, only weight and slant. Inline `code` and a
[link](https://example.org) do carry one.

> A quote takes the string colour.

- a list marker takes the keyword colour
- so does this one


```
# every coloured kind in one cell, so one screenshot judges the whole palette
import math


def area(radius):
    """A docstring is a string."""
    return math.pi * radius ** 2


count = 42
label = f'{count} circles of area {area(1.5):.3f}'
print(label)
```
